## 1. NumPy Operations

```python
# TODO: Create a NumPy array and perform vectorized operations
prices = np.array([100,102,98,105,110]) #create a numpy vector, 5 days stock price
returns = (prices[1:]-prices[:-1]) / prices[1:] #vectorized calculation: prices[1:] is 'from the 2nd value to the last value', prices[:-1]
#is 'from the 1st value till the 2nd last value', essentially subtract the 2nd value from the 1st one for all values in the vector.
print("Prices:", prices)
print("Elementwise % change:", returns)
```

## 2. Compare Loop vs Vectorization

```python
# TODO: Compare speed using %timeit
big_array = np.arange(1_000_000) #create a vector from 0 to 999,999

%timeit [x * 2 for x in big_array] #essentially a for loop, for every value in big_array, times it by 2 and gives out a new list
%timeit big_array * 2 #vectorization operation: directly apply multiplication to the entire vector, one and done calculation 
```

## 5. Reusable Function 

```python
def get_summary_stats(df, group_col='category', value_col='value'):
    """
    Group by the specified column and calculate the statistical summaries of the numeric column.

    parameters:
        df (pd.DataFrame): dataframe input
        group_col (str): names of the column for grouping，default: 'category'
        value_col (str): column used for calculation，default: 'value'

    return:
        pd.DataFrame: The statistical summaries for each group (mean, standard deviation, maximum and minimum values, count)
    """
    stats = df.groupby(group_col)[value_col].agg(['mean', 'std', 'min', 'max', 'count']).reset_index()
    return stats
```

```python
result = get_summary_stats(df)
result
```
